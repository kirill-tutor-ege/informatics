### Краткая суть задачи

**Найти количество способов получить из числа A число B, используя команды, приведенные в задании**

*Пример задачи:*<br>
Исполнитель преобразует число на экране. У исполнителя есть три команды, которые обозначены латинскими буквами:<br>
A. Вычесть 1<br>
B. Вычесть 4<br>
C. Найти целую часть от деления на 3<br>
Программа для исполнителя – это последовательность команд.
Сколько существует программ, для которых при исходном числе 19 результатом является 2, при этом траектория вычислений не содержит числа 7 и содержит 13?

>Задача решается шаблонным рекурсивным алгоритмом, ниже приведены основные разновидности с примерами

##### Вариант 1. Найти количество путей из меньшего числа в большее, без дополнительных усложнений

```python
def f(n, finish):
    if n == finish:
        return 1
    if n > finish:
        return 0
    return f(команда1, finish) + f(команда2, finish) ...

print(f(старт, финиш))
```

*Пример:*
У исполнителя Увеличитель две команды, которым присвоены номера.<br>
1.  Прибавь 2.<br>
2.  Умножь на 3.<br>
Первая из них увеличивает число на экране на 2, вторая  — умножает его на 3.<br>
Сколько есть программ, которые число 1 преобразуют в число 31?<br>

*Решение:*

````python
def f(n, finish):
    if n > finish:
        return 0
    if n == finish:
        return 1
    return f(n + 2, finish) + f(n * 3, finish)

print(f(1, 31))
````

Если нужно найти из большего числа в меньшее, то вторая строчка 
```python 
if n > finish:
```
меняется на 
```python
if n < finish:
```

##### Вариант 2. Найти количество путей из меньшего числа в большее с избегаемым этапом (или этапами)

Избегаемый этап — просто дополнительный базовый случай рекурсии, при котором мы возвращаем 0

```python
def f(n, finish):
    if n == finish:
        return 1
    if n > finish or n == избегаемое_число1 or n == избегаемое_число2 ...:
        return 0
    return f(команда1, finish) + f(команда2, finish) ...

print(f(старт, финиш))
```

*Пример:*<br>
Исполнитель НечетМ преобразует число на экране. У исполнителя НечетМ две команды, которым присвоены номера.<br>
1.  Прибавь 1.<br>
2.  Сделай нечётное.<br>
Первая из этих команд увеличивает число x на экране на 1, вторая переводит число x в число 2x + 1.<br>
Сколько существует таких программ, которые число 1 преобразуют в число 27, причём траектория вычислений не содержит число 26? <br>

````python
def f(n, finish):
    if n == finish:
        return 1
    if n > finish or n == 26:
        return 0
    return f(n + 1, finish) + f(n * 2 + 1, finish) ...

print(f(1, 27))
````

##### Вариант 3. Найти количество путей из меньшего числа в большее с обязательным этапом (или этапами)

Когда в задаче появляется обязательный этап (этапы), мы используем правило перемножения из комбинаторики, и находим пути A → B → C как A → B × A → C

```python
def f(n, finish):
    if n == finish:
        return 1
    if n > finish:
        return 0
    return f(команда1, finish) + f(команда2, finish) ...

print(f(старт, обязательный_этап1) * f(обязательный_этап1, обязательный_этап2) * ... * f(обязательный_этапN, финиш))
```

*Пример:*<br>
Исполнитель А16 преобразует число, записанное на экране. У исполнителя есть три команды, которым присвоены номера.<br>
1.  Прибавить 1.<br>
2.  Прибавить 2.<br>
3.  Умножить на 2.<br>
Сколько существует таких программ, которые исходное число 3 преобразуют в число 12 и при этом траектория вычислений программы содержит число 10?<br>

````python
def f(n, finish):
    if n == finish:
        return 1
    if n > finish:
        return 0
    return f(n + 1, finish) + f(n + 2, finish) + f(n * 2, finish)

print(f(3, 10) * f(10, 12))
````

##### Вариант 4 (актуальный). Найти количество путей из меньшего числа в большее с обязательным  и избегаемым этапом (или этапами)

Комбинируем подходы из двух предыдущих примеров

```python
def f(n, finish):
    if n == finish:
        return 1
    if n > finish  or n == избегаемое_число1 or n == избегаемое_число2 ...:
        return 0
    return f(команда1, finish) + f(команда2, finish) ...

print(f(старт, обязательный_этап1) * f(обязательный_этап1, обязательный_этап2) * ... * f(обязательный_этапN, финиш))
```

*Пример:*<br>
Исполнитель преобразует число на экране. У исполнителя есть три команды, которые обозначены латинскими буквами:<br>
A. Вычесть 1<br>
B. Вычесть 4<br>
C. Найти целую часть от деления на 3<br>
Сколько существует программ, для которых при исходном числе 19 результатом является 2, при этом траектория вычислений не содержит числа 7 и содержит 13?<br>

````python
def f(n, finish):
    if n == finish:
        return 1
    if n < finish or n == 7:
        return 0
    return f(n - 1, finish) + f(n - 4, finish) + f(n // 3, finish)

print(f(19, 13) * f(13, 2))
````

##### Вариант 5 (Сложнее ЕГЭ). Найти количество путей из меньшего числа в большее с ограничениями на количество или порядок следования команд

В данном усложнении самое главное — к моменту проверки подходящего пути собрать полную последовательность команд, которой оно было получено, это делается следующим образом

Команда А. Прибавить 1
Команда Б. Прибавить 2

```python
def f(n, finish, commands):
    if n == finish:
        if <какая-то проверка для переменной commands>:
            return 1
        else:
            return 0
    if n > finish:
        return 0
    return f(n + 1, finish, commands + 'A') + f(n + 2, finish, commands + 'B')
```

*Пример:*<br>
Исполнитель преобразует число на экране. У исполнителя есть три команды, которые обозначены латинскими буквами:<br>
А. Вычесть 1<br>
В. Прибавить 5<br>
C. Умножить на 2<br>
Сколько существует программ, для которых при исходном числе 5 результатом является число 34 и при этом команда A не повторяется более двух раз подряд?<br>

````python
def f(n, finish, commands):
    if n == finish:
        return 1
    if n - 2 > finish or "AAA" in commands: # Из-за того что среди команд B и C, команда А вычитает единицу
        return 0
    return f(n - 1, finish, commands + 'A') + f(n + 5, finish, commands + 'B') + f(n * 2, finish, commands + 'C')

print(f(5, 34, ''))
````

**Задание №1**<br>
Алгоритм вычисления значения функции F(n), где n – целое число, задан следующими соотношениями:<br>

F(n) = 1, при n = 1;<br>
F(n) = 2 × n + F(n − 1), если n > 1.<br>

Чему равен квадрат суммы цифр значения функции F(57693)?<br>

*Источник: КЕГЭ. Л. Шастин. Задание №14337.*<br>
<details>
  <summary><b>Показать ответ</b></summary>
  Ответ: 2209
</details>

```python
from functools import *

def digits_sum(n):
    return sum(int(x) for x in str(n))

@lru_cache(None)
def F(n):
    if n == 1:
        return 1
    if n > 1:
        return 2 * n + F(n - 1)

for x in range(57693 + 1):
    F(x)

square_of_digits_sum = digits_sum(F(57693)) ** 2
print(square_of_digits_sum)
```

**Задание №2**<br>
Алгоритм вычисления значения функции F(n), где n – натуральное число, задан следующими соотношениями:<br>

F(n) = 1, при n = 1;<br>
F(n) = (4n + 7) × F(n − 1) + 16, если n > 1.<br>

Определите количество таких целых k, что 10¹² ⩽ k ⩽ 2 × 10¹²  и F(k) чётно.<br>

*Источник: КЕГЭ. М. Попков. Задание №13301.*<br>
<details>
  <summary><b>Показать ответ</b></summary>
  Ответ: 2209
</details>

```python
from functools import *

def digits_sum(n):
    return sum(int(x) for x in str(n))

@lru_cache(None)
def F(n):
    if n == 1:
        return 1
    if n > 1:
        return (4 * n + 7) * F(n - 1) + 16

for x in range(2 * 10**12):
    F(x)
```
